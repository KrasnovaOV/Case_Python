Отберите обращения по продуктам **«GoPro Hero»** или **«Dell XPS»**,
где приоритет **«High»** или **«Critical»**.

```python
# Ваше решение:
# Решение задания 2.1 — Проблемные обращения
# Отбор обращений с оценкой удовлетворенности ниже 3.0.
# Фильтрация
df_low_satisfaction = df[df['Customer Satisfaction Rating'] < 3.0]

# Сохранение количества
count_low_sat = len(df_low_satisfaction)

print(f"Задание 2.1:")
print(f"Количество обращений с оценкой ниже 3.0: {count_low_sat}")
print("Первые 5 строк проблемных обращений:")
display(df_low_satisfaction.head())
```

## Задание 3: Вычисляемые столбцы

Для оценки качества работы службы поддержки нужны дополнительные показатели.

**Задача:**

### 3.1. Признак низкой удовлетворённости
Создайте новый столбец `is_low_satisfaction`, который равен `True`,
если `Customer Satisfaction Rating` меньше 3.0, и `False` в противном случае.

### 3.2. Категория возраста клиента
Создайте столбец `age_group` по правилу:
- Если `Customer Age` < 30 → значение **«Молодой»**
- Если `Customer Age` < 50 → значение **«Средний»**
- Если `Customer Age` >= 50 → значение **«Старший»**



```python
# Ваше решение:
# --- Задание 3.1. Признак низкой удовлетворённости ---
# Создаем булев столбец.
# Если рейтинг меньше 3.0 -> True, иначе False. Для NaN результат будет False.
df['is_low_satisfaction'] = df['Customer Satisfaction Rating'] < 3.0
```

```python
# --- Задание 3.2. Категория возраста клиента ---
# Используем pd.cut для автоматического разбиения на интервалы
bins = [0, 30, 50, float('inf')]
labels = ['Молодой', 'Средний', 'Старший']

# Параметр right=False делает интервалы вида [от, до), то есть 30 попадет в "Средний", а не "Молодой"
df['age_group'] = pd.cut(
    df['Customer Age'],
    bins=bins,
    labels=labels,
    right=False
)

# --- Вывод результатов для проверки ---
print("=== РЕЗУЛЬТАТЫ ЗАДАНИЯ 3 ===\n")

print("--- 3.1. Столбец is_low_satisfaction (первые 10 строк) ---")
display(df[['Ticket ID', 'Customer Satisfaction Rating', 'is_low_satisfaction']].head(10))

print("\n--- 3.2. Столбец age_group (первые 10 строк) ---")
display(df[['Ticket ID', 'Customer Age', 'age_group']].head(10))

print("\n--- Статистика по новым признакам ---")
print(f"Количество клиентов с низкой оценкой (< 3.0): {df['is_low_satisfaction'].sum()}")
print("Распределение по возрастным группам:")
print(df['age_group'].value_counts(dropna=False))
```

### ⭐ Задание 6: Проблемный продукт (задание со звёздочкой)

*Это задание не является обязательным, но поможет вам лучше освоить материал.*

Руководитель подозревает, что один из продуктов
генерирует больше всего жалоб с низкими оценками. Нужно проверить эту гипотезу.

**Задача:**
1. Определите продукт, на который поступило **больше всего** обращений с оценкой ниже 3.0.
2. Посчитайте среднюю оценку удовлетворённости для этого продукта.
3. Сравните её со средней оценкой по всем продуктам.
4. Сделайте вывод — подтверждается ли гипотеза руководителя?



```python
# Ваше решение:
# --- Шаг 1: Определение проблемного продукта ---
# Фильтруем только обращения с низкой оценкой (< 3.0)
df_low = df[df['Customer Satisfaction Rating'] < 3.0]

# Находим продукт-лидер по количеству таких обращений
problem_product = df_low['Product Purchased'].value_counts().idxmax()
problem_product_count = df_low['Product Purchased'].value_counts().max()

# --- Шаг 2: Расчет средних оценок ---
# Средняя оценка для этого конкретного продукта (по ВСЕМ его тикетам)
avg_sat_problem_product = df[df['Product Purchased'] == problem_product]['Customer Satisfaction Rating'].mean()

# Средняя оценка по всем продуктам в базе (базовая линия)
overall_avg_sat = df['Customer Satisfaction Rating'].mean()

# --- Шаг 3: Проверка гипотезы ---
is_hypothesis_confirmed = avg_sat_problem_product < overall_avg_sat

# --- Вывод результатов ---
print(f"1. Продукт с наибольшим количеством жалоб (< 3.0):")
print(f"   -> {problem_product} ({problem_product_count} обращений)")
print("-" * 50)

print(f"2. Средняя оценка удовлетворённости для «{problem_product}» (по всем его тикетам):")
print(f"   -> {avg_sat_problem_product:.2f}")
print("-" * 50)

print(f"3. Средняя оценка удовлетворённости по всем продуктам (базовая линия):")
print(f"   -> {overall_avg_sat:.2f}")
print("-" * 50)

print("4. Вывод:")
if is_hypothesis_confirmed:
    print(f"   Гипотеза руководителя ПОДТВЕРЖДАЕТСЯ. Средняя оценка продукта «{problem_product}» ({avg_sat_problem_product:.2f}) ниже общей средней по базе ({overall_avg_sat:.2f}).")
    print("   Этот продукт не только генерирует больше всего негативных отзывов, но и имеет системно низкое качество клиентского опыта.")
else:
    print(f"   Гипотеза руководителя НЕ ПОДТВЕРЖДАЕТСЯ. Несмотря на большое количество жалоб, средняя оценка «{problem_product}» ({avg_sat_problem_product:.2f}) не ниже средней по базе ({overall_avg_sat:.2f}).")
    print("   Проблема, скорее всего, заключается в массовости продукта, а не в его качестве.")
```
