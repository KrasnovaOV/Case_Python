```python
# Решение для всех заданий

employees_data = [
    {'name': 'Анна Смирнова', 'department': 'Продажи', 'sales': 1500000, 'deals_count': 45, 'satisfaction': 4.8},
    {'name': 'Игорь Петров', 'department': 'Продажи', 'sales': 2300000, 'deals_count': 62, 'satisfaction': 4.5},
    {'name': 'Елена Козлова', 'department': 'Маркетинг', 'sales': 800000, 'deals_count': 15, 'satisfaction': 4.9},
    {'name': 'Дмитрий Иванов', 'department': 'Продажи', 'sales': 950000, 'deals_count': 28, 'satisfaction': 4.2},
    {'name': 'Мария Сидорова', 'department': 'Маркетинг', 'sales': 1200000, 'deals_count': 22, 'satisfaction': 4.7},
    {'name': 'Алексей Морозов', 'department': 'Продажи', 'sales': 3100000, 'deals_count': 78, 'satisfaction': 4.6},
    {'name': 'Ольга Волкова', 'department': 'Поддержка', 'sales': 0, 'deals_count': 0, 'satisfaction': 4.9},
    {'name': 'Сергей Кузнецов', 'department': 'Продажи', 'sales': 1800000, 'deals_count': 51, 'satisfaction': 4.4},
    {'name': 'Наталья Белова', 'department': 'Маркетинг', 'sales': 950000, 'deals_count': 18, 'satisfaction': 4.3},
    {'name': 'Павел Новиков', 'department': 'Поддержка', 'sales': 0, 'deals_count': 0, 'satisfaction': 4.8}
]

# Вспомогательные функции

def get_overall_metrics(employees):
    total_sales = sum(e['sales'] for e in employees)
    total_deals = sum(e['deals_count'] for e in employees)
    avg_check = total_sales / total_deals if total_deals else 0
    return {
        'total_sales': total_sales,
        'total_deals': total_deals,
        'avg_check': avg_check
    }


def get_department_sales(employees):
    dept_sales = {}
    for emp in employees:
        d = emp['department']
        dept_sales[d] = dept_sales.get(d, 0) + emp['sales']
    # Сортируем по убыванию продаж для красивого вывода
    sorted_dept = dict(sorted(dept_sales.items(), key=lambda x: x[1], reverse=True))
    return sorted_dept


def get_top_performers_by_sales(employees, top_n=3):
    sorted_emps = sorted(employees, key=lambda x: x['sales'], reverse=True)
    return [{'name': e['name'], 'sales': e['sales']} for e in sorted_emps[:top_n]]


def get_candidates_for_training(employees):
    return [
        emp.copy() # Создаем поверхностную копию словаря сотрудника
        for emp in employees
        if emp['deals_count'] == 0 or emp['satisfaction'] < 4.5
    ]


def get_above_average_performers(employees):
    metrics = get_overall_metrics(employees)
    avg_sales = metrics['total_sales'] / len(employees) if employees else 0
    return [
        {'name': e['name'], 'sales': e['sales']} for e in employees if e['sales'] > avg_sales
    ]


def calculate_bonuses(employees, total_bonus_fund):
    result = []
    metrics = get_overall_metrics(employees)
    total_sales = metrics['total_sales']
    for emp in employees:
        bonus = (emp['sales'] / total_sales * total_bonus_fund) if total_sales else 0
        if emp['satisfaction'] < 4.0:
            penalty = bonus / 2
            bonus -= penalty
        new_emp = emp.copy()
        new_emp['bonus'] = round(bonus, 2)
        result.append(new_emp)
    return result


def generate_report(data):
    print("=== ОТЧЁТ ОБ ЭФФЕКТИВНОСТИ СОТРУДНИКОВ ===\n")
    # Общие показатели
    overall = get_overall_metrics(data)
    print(f"Общие показатели:\n"
          f"- Общая сумма продаж: {overall['total_sales']:,} руб.\n"
          f"- Общее количество сделок: {overall['total_deals']}\n"
          f"- Средняя сумма сделки: {overall['avg_check']:,.2f} руб.")

    # Аналитика по отделам
    dept_sales = get_department_sales(data)
    print("\nАналитика по отделам:")
    for dept, sales in dept_sales.items():
        print(f"- {dept}: {sales:,} руб.")

    # Кандидаты на премию
    top3 = get_top_performers_by_sales(data)
    names = ', '.join([f"{x['name']} ({x['sales']:,} руб.)" for x in top3])
    print("\nТоп-3 сотрудника по сумме продаж:")
    for item in top3:
        print(f"- {item['name']}: {item['sales']:,} руб.")

    # Кандидаты на обучение
    training = get_candidates_for_training(data)
    print("\nКандидаты на обучение (сделок = 0 или удовлетворённость < 4.5):")
    if not training:
        print("- Нет кандидатов на обучение в этом квартале.")
    else:
        for emp in training:
            reason = []
            if emp['deals_count'] == 0:
                reason.append('нет закрытых сделок')
            if emp['satisfaction'] < 4.5:
                reason.append(f'удовлетворённость {emp['satisfaction']}')
            reasons_str = ", ".join(reason)
            print(f"- {emp['name']} ({emp['department']}): {reasons_str}")

    # Задание 2: Выше среднего
    above_avg = get_above_average_performers(data)
    print("\nЗадание 2: Сотрудники с продажами выше среднего:")
    if not above_avg:
        print("- Таких сотрудников нет.")
    else:
        for item in above_avg:
            print(f"- {item['name']}: {item['sales']:,} руб.")

    # Задание 3: Расчёт бонусов
    bonus_fund = 500_000
    bonused = calculate_bonuses(data, bonus_fund)
    print("\nЗадание 3: Распределение премиального фонда ({:,.0f} руб.):".format(bonus_fund))
    for emp in bonused:
        print(f"- {emp['name']}: {emp['bonus']:,.2f} руб.")


if __name__ == '__main__':
    generate_report(employees_data)
```
