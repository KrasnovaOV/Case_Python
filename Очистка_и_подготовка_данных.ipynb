## **Задание 4. Замена и удаление выбросов**

Числовые столбцы могут содержать экстремальные значения, которые искажают статистику. Найдите их и обработайте.

**Задачи:**

### 4.1. Анализ столбца `tip`

1. Выведите статистику для столбца `tip`.
2. Найдите выбросы методом IQR:
   - Рассчитайте Q1, Q3 и IQR.
   - Определите нижнюю и верхнюю границы: `Q1 − 1.5 × IQR` и `Q3 + 1.5 × IQR`.
   - Подсчитайте, сколько значений выходит за границы.
3. Обрежьте выбросы с помощью `.clip(lower=lower_bound, upper=upper_bound)`.

### 4.2. Анализ столбца `customer_age`

1. Проверьте, есть ли значения возраста меньше 14 или больше 100 — это заведомо некорректные данные.
2. Удалите строки с такими значениями.
3. Выведите `df["customer_age"].describe()` до и после очистки. Сравните min, max и среднее.
   
**Вопрос:**

```python
# Ваше решение:
import pandas as pd

# --- ЗАДАНИЕ 4. ПОИСК И ОБРАБОТКА ВЫБРОСОВ ---
print("ЗАДАНИЕ 4: ОБРАБОТКА ВЫБРОСОВ")
print("=" * 50)

# Фиксируем состояние до изменений для итогового отчета
tip_before = df['tip'].copy()
age_before = df['customer_age'].copy()
initial_rows = len(df)


# --- Задача 4.1: Анализ и очистка столбца 'tip' ---
print("\nЗАДАЧА 4.1: Обработка выбросов в 'tip'")

# Шаг 1: Замена отрицательных значений на 0 (бизнес-логика: чаевые не могут быть штрафом < 0)
negative_tips_count = (df['tip'] < 0).sum()
if negative_tips_count > 0:
    print(f"Обнаружено {negative_tips_count} отрицательных значений чаевых. Они заменены на 0.")
    df.loc[df['tip'] < 0, 'tip'] = 0

print("Статистика 'tip' ДО очистки статистических выбросов:")
display(df['tip'].describe())

# Шаг 2: Расчет границ IQR только по валидным данным
Q1 = df['tip'].quantile(0.25)
Q3 = df['tip'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = max(Q1 - 1.5 * IQR, 0) # Нижняя граница принудительно >= 0
upper_bound = Q3 + 1.5 * IQR

print(f"\nРасчеты по методу IQR:")
print(f"Нижняя граница: {lower_bound}")
print(f"Верхняя граница: {upper_bound}")

outliers_low = df[df['tip'] < lower_bound]['tip'].count()
outliers_high = df[df['tip'] > upper_bound]['tip'].count()
total_outliers = outliers_low + outliers_high
print(f"Найдено статистических выбросов ниже границы: {outliers_low}")
print(f"Найдено статистических выбросов выше границы: {outliers_high}")

# Шаг 3: Обрезка выбросов (все данные уже неотрицательные благодаря шагу 1)
# Если после обрезки вдруг появятся NaN (например, если они были изначально), заполним их медианой
tip_median = df['tip'].median()
df['tip'] = df['tip'].clip(lower=lower_bound, upper=upper_bound).fillna(tip_median)

print("\nСтатистика 'tip' ПОСЛЕ полной очистки:")
display(df['tip'].describe())
print("-" * 50)


# --- Задача 4.2: Анализ и очистка столбца 'customer_age' ---
print("\nЗАДАЧА 4.2: Очистка аномалий в 'customer_age'")

print("Описательная статистика 'customer_age' ДО удаления строк:")
age_desc_before = df['customer_age'].describe()
display(age_desc_before)

# Фильтрация заведомо невозможных значений (<14 или >100)
mask_valid_ages = (df['customer_age'] >= 14) & (df['customer_age'] <= 100)
pre_filter_count = len(df)
df = df[mask_valid_ages].copy()
post_filter_count = len(df)
removed_count = pre_filter_count - post_filter_count

print(f"\nУдалено строк с некорректным возрастом (<14 или >100): {removed_count}")

print("\nОписательная статистика 'customer_age' ПОСЛЕ удаления строк:")
age_desc_after = df['customer_age'].describe()
display(age_desc_after)

# Сравнение ключевых метрик возраста
comparison_data = {
    'Метрика': ['min', 'max', 'mean'],
    'До очистки': [
        age_desc_before['min'],
        age_desc_before['max'],
        round(age_desc_before['mean'], 2)
    ],
    'После очистки': [
        age_desc_after['min'],
        age_desc_after['max'],
        round(age_desc_after['mean'], 2)
    ]
}
display(pd.DataFrame(comparison_data))
print("-" * 50)


# --- Ответ на вопрос задания ---
print("\nОТВЕТ НА ВОПРОС К ЗАДАНИЮ 4.1:")
tip_after = df['tip']

old_mean = tip_before.mean()
new_mean = tip_after.mean()
change_mean = new_mean - old_mean

old_std = tip_before.std()
new_std = tip_after.std()
change_std = new_std - old_std

print(f"Среднее значение 'tip' изменилось с {old_mean:.2f} до {new_mean:.2f} ({change_mean:+.2f}).")
print(f"Стандартное отклонение 'tip' изменилось с {old_std:.2f} до {new_std:.2f} ({change_std:+.2f}).")

if change_mean > 0:
    print("-> Среднее увеличилось. Замена отрицательных значений на 0 подняла среднее вверх сильнее, чем его опустила обрезка высоких выбросов.")
else:
    print("-> Среднее уменьшилось или осталось прежним. Удаление экстремально высоких чаевых оказало большее влияние на снижение среднего.")

if new_std < old_std:
    print("-> Стандартное отклонение уменьшилось. Данные стали более однородными (сжались к центру).")
```

```python
# Ваше решение:
# Фиксируем размер датасета до начала чистки
rows_before_integrity = df.shape[0]
print(f"Размер датасета на начало задания: {rows_before_integrity} строк")
print("=" * 50)


# --- Задача 1: Проверка tip <= price ---
print("\nЗАДАЧА 1: Логическая проверка 'tip' и 'price'")

# Создаем маску некорректных строк (строгое неравенство, так как чаевые могут быть равны цене "оставить всё")
mask_tip_gt_price = df['tip'] > df['price']
invalid_tips_count = mask_tip_gt_price.sum()

if invalid_tips_count > 0:
    print(f"Обнаружено {invalid_tips_count} заказов, где чаевые превышают стоимость.")
    # Выведем эти строки для анализа
    display(df[mask_tip_gt_price][['order_id', 'price', 'tip']])

    # Удаляем их из основного датафрейма
    df = df[~mask_tip_gt_price].copy()
    print(f"Удалено строк по условию tip > price: {invalid_tips_count}")
else:
    print("Ошибок не обнаружено: ни в одном заказе чаевые не превышают цену.")
print("-" * 50)


# --- Задача 2: Проверка quantity ---
print("\nЗАДАЧА 2: Проверка столбца 'quantity'")

# Сначала убедимся, что колонка точно числовая (на случай, если Step 2 был пропущен)
df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce')

# Маска для поиска нулевых, отрицательных или дробных значений
mask_invalid_quantity = (df['quantity'] <= 0) | (df['quantity'] % 1 != 0)
invalid_qty_count = mask_invalid_quantity.sum()

if invalid_qty_count > 0:
    print(f"Обнаружено {invalid_qty_count} строк с некорректным количеством порций (<= 0 или не целое).")
    display(df[mask_invalid_quantity][['order_id', 'drink', 'quantity']])

    df = df[~mask_invalid_quantity].copy()
    print(f"Удалено строк по условию quantity: {invalid_qty_count}")
else:
    print("Ошибок не обнаружено: все значения quantity положительные и целые.")
print("-" * 50)


# --- Задача 3: Проверка payment_method ---
print("\nЗАДАЧА 3: Согласованность 'payment_method'")

allowed_methods = ['Cash', 'Card', 'App']

# Находим уникальные значения, которых нет в нашем белом списке
unique_methods = df['payment_method'].unique()
invalid_methods = [m for m in unique_methods if m not in allowed_methods]

if invalid_methods:
    print(f"Обнаружены недопустимые способы оплаты: {invalid_methods}")
    # Покажем распределение этих мусорных значений
    display(df[df['payment_method'].isin(invalid_methods)]['payment_method'].value_counts())

    # РЕШЕНИЕ: Заменим неизвестные методы на категорию 'App' вместо удаления строк,
    # чтобы сохранить информацию о факте транзакции.
    df['payment_method'] = df['payment_method'].apply(lambda x: x if x in allowed_methods else 'App')
    print("Недопустимые значения заменены на 'App'.")
else:
    print("Все способы оплаты соответствуют списку: Cash, Card, App.")
print("-" * 50)


# --- Итоговые проверки ---
print("\nИТОГОВЫЕ ПРОВЕРКИ:")
final_rows = df.shape[0]
removed_total = rows_before_integrity - final_rows

print(f"Итоговый размер датасета: {df.shape[0]} строк x {df.shape[1]} столбцов")
print(f"Всего удалено строк за это задание: {removed_total}")
print("-" * 50)

print("\nИнформация о типах данных после всех проверок (df.info()):")
df.info(memory_usage='deep', verbose=False)
```
