```python
import pandas as pd

class SalesAnalyzer:
    def __init__(self, file_path):
        """
        Инициализация: загрузка данных из CSV-файла.
        Приводит названия колонок к нижнему регистру, ищет 'quantity' и 'price',
        добавляет столбец 'revenue'. При ошибке выводит сообщение.
        """
        self.__data = None
        try:
            df = pd.read_csv(file_path)
            # Нормализуем имена колонок
            df.columns = [str(col).strip().lower() for col in df.columns]

            # Поиск нужных колонок
            quantity_col = next((col for col in df.columns if 'quantity' in col), None)
            price_col = next((col for col in df.columns if 'price' in col), None)

            # Расчет выручки, если возможно
            if quantity_col and price_col:
                df['revenue'] = df[quantity_col] * df[price_col]
            else:
                print("Предупреждение: не найдены колонки 'quantity' и/или 'price'. Столбец 'revenue' не создан.")

            self.__data = df
            print(f"Данные успешно загружены. Размер: {df.shape}.")
        except Exception as e:
            print(f"Ошибка при загрузке файла: {e}")

    def get_basic_info(self):
        """Вывод базовой информации о DataFrame."""
        if self.__data is not None:
            print(f"Размер данных: {self.__data.shape}")
            print(f"Колонки: {list(self.__data.columns)}")
            print("Типы данных:\n", self.__data.dtypes)
            print("Первые 3 строки:\n", self.__data.head(3))
        else:
            print("Данные не загружены.")

    def get_revenue_by_category(self):
        """
        Возвращает Series с суммарной выручкой по категориям.
        Ищет колонку с 'category' или 'product_line'.
        """
        if self.__data is None or 'revenue' not in self.__data.columns:
            return None

        category_col = next(
            (col for col in self.__data.columns if 'category' in col or 'product_line' in col),
            None
        )
        if category_col:
            result = (
                self.__data.groupby(category_col)['revenue']
                          .sum()
                          .sort_values(ascending=False)
            )
            return result
        else:
            print("Колонка категории ('category', 'product_line') не найдена.")
            return None

    def get_top_products(self, n=5):
        """
        Возвращает топ-N товаров по количеству продаж.
        Ищет колонки с 'product' и 'quantity'.
        """
        if self.__data is None:
            return None

        product_col = next((col for col in self.__data.columns if 'product' in col), None)
        quantity_col = next((col for col in self.__data.columns if 'quantity' in col), None)

        if product_col and quantity_col:
            top_n = (
                self.__data.groupby(product_col)[quantity_col]
                          .sum()
                          .sort_values(ascending=False)
                          .head(n)
            )
            return top_n
        else:
            print("Не найдены колонки для анализа топ товаров ('product', 'quantity').")
            return None

    def display_summary(self):
        """Выводит общую выручку, топ-3 категории и самый популярный товар."""
        rev_by_cat = self.get_revenue_by_category()
        total_rev = self.__data['revenue'].sum() if self.__data is not None and 'revenue' in self.__data.columns else None
        top_prod = self.get_top_products(n=1)  # Самый популярный товар

        print("\n=== Сводный отчёт ===")
        print(f"Общая выручка: {total_rev:,.2f}" if total_rev is not None else "Общая выручка: нет данных")

        if isinstance(rev_by_cat, pd.Series) and len(rev_by_cat) > 0:
            print("Топ-3 категории по выручке:")
            for i, (cat, val) in enumerate(rev_by_cat.head(3).items(), start=1):
                print(f"{i}. {cat}: {val:,.2f}")
        else:
            print("Топ-3 категории: нет данных")

        if isinstance(top_prod, pd.Series) and len(top_prod) > 0:
            most_pop_product, qty = top_prod.index[0], top_prod.iloc[0]
            print(f"Самый популярный товар: '{most_pop_product}' (продано: {qty})")
        else:
            print("Самый популярный товар: нет данных")
```

```python
# Создаём экземпляр анализатора, передавая ID файла
analyzer = SalesAnalyzer('sales_data.csv')

print("\n=== Базовая информация ===")
analyzer.get_basic_info()

print("\n=== Выручка по категориям ===")
rev_by_cat = analyzer.get_revenue_by_category()
if rev_by_cat is not None:
    print(rev_by_cat)

print("\n=== Топ товаров по количеству продаж ===")
top_prod = analyzer.get_top_products(n=5)
if top_prod is not None:
    print(top_prod)

print("\n=== Итоговый отчёт ===")
analyzer.display_summary()
```
